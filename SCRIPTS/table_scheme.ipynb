```python
import pandas as pd
import json
import os
```

```python
for file in os.listdir(r'C:\Users\kgathola.puka\OneDrive - MSC\Documents\GitHub\RCP(test)\SPEED CHATBOT PROJECT\DATA\Database Tables'):
    if file.endswith('.xlsx') and file.startswith('SQL'):
        print(f'File being processed: {file}')
        df = pd.read_excel(os.path.join(r'C:\Users\kgathola.puka\OneDrive - MSC\Documents\GitHub\RCP(test)\SPEED CHATBOT PROJECT\DATA\Database Tables', file))
        df["is_primary_key"] = df["is_primary_key"].astype(str).str.upper() == "TRUE"
        df["is_foreign_key"] = df["is_foreign_key"].astype(str).str.upper() == "TRUE"
        table_name = df["Table_name"].iloc[0]
        table_description = df["Table_Description"].iloc[0]


        primary_key = df["Primary_Key"].iloc[0] if not df["Primary_Key"].isna().all() else None

        columns = []

        for _, row in df.iterrows():
            column = {
                "name": row["Column_Name"],
                "description": row["Column_Description"],
                "type_sql_server": row["SQL_Server_Data_Type"],
                "type_oracle": row["Oracle_Data_Type"],
                "is_primary_key": bool(row["is_primary_key"]),
                "is_foreign_key": bool(row["is_foreign_key"]),
                "references_table": None if pd.isna(row["references_table"]) else row["references_table"],
                "references_column": None if pd.isna(row["references_column"]) else row["references_column"]
            }

            columns.append(column)


        schema = {
            "table_name": table_name,
            "description": table_description,
            "primary_key": primary_key,
            "columns": columns
        }
        with open(f"C:\\Users\\kgathola.puka\\OneDrive - MSC\\Documents\\GitHub\\RCP(test)\\SPEED CHATBOT PROJECT\\DATA\\Database Tables\\{table_name}.json", "w", encoding="utf-8") as f:
            json.dump(schema, f, indent=2)
            print(" saved")
```
